Multi-physics simulation of convective heat extraction and non-conductive barrier torque transfer.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# Configure Matplotlib math rendering
mpl.rcParams['text.usetex'] = False
mpl.rcParams['mathtext.fontset'] = 'cm'

# Domain Setup
time_steps = 300
delta_T_range = np.linspace(10, 350, time_steps)
rpm_range = np.linspace(500, 10000, time_steps)

# Binary He-Xe Mixture Parameters
cp_he_xe = 850.0      # J/(kg*K)
m_dot_he_xe = 0.28    # kg/s
Q_he_xe_kW = (m_dot_he_xe * cp_he_xe * delta_T_range) / 1000.0

# Baseline Pure Helium
cp_pure_he = 5193.0   # J/(kg*K)
m_dot_pure_he = 0.025 # kg/s
Q_pure_he_kW = (m_dot_pure_he * cp_pure_he * delta_T_range) / 1000.0

# Magnetic Drive & Si3N4 Ceramic Barrier
Torque_Nm = 4.5 * np.ones_like(rpm_range)
P_eddy_Si3N4 = np.zeros_like(rpm_range)  # Zero electrical conductivity

# Plotting
fig, axs = plt.subplots(2, 1, figsize=(9, 7))

# Subplot 1: Heat Extraction
axs[0].plot(delta_T_range, Q_pure_he_kW, 'g--', linewidth=2, label='Baseline Pure Helium (m_dot=0.025 kg/s)')
axs[0].plot(delta_T_range, Q_he_xe_kW, 'b-', linewidth=2.5, label='Optimized He-Xe Mixture (m_dot=0.28 kg/s)')
axs[0].set_ylabel('Heat Extraction Rate (kW)')
axs[0].set_xlabel('Thermal Differential Delta T (C)')
axs[0].set_title('Convective Heat Extraction in UHV Thermal Loop')
axs[0].grid(True, ls='--')
axs[0].legend(loc='upper left')

# Subplot 2: Torque & Eddy Loss (Dual Axis)
ax2 = axs[1].twinx()
line1 = axs[1].plot(rpm_range, Torque_Nm, 'b-', linewidth=2, label='Magnetic Coupling Torque (N m)')
line2 = ax2.plot(rpm_range, P_eddy_Si3N4, 'r-', linewidth=2, label='Si3N4 Ceramic Barrier Eddy Loss (W)')

axs[1].set_xlabel('Pump Rotational Velocity (RPM)')
axs[1].set_ylabel('Torque Capacity (N m)', color='b')
ax2.set_ylabel('Barrier Power Loss (W)', color='r')
ax2.set_ylim(-0.1, 1.0)
axs[1].grid(True, ls='--')

# Combine handles and labels from both twin axes
lines = line1 + line2
labels = [l.get_label() for l in lines]
axs[1].legend(lines, labels, loc='center right')

plt.tight_layout()
plt.show()
```
